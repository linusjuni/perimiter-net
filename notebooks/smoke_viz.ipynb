```python
import sys
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from pathlib import Path

# Hardcoded path - update this to your results directory
PATH_TO_RESULTS = "/work3/s225224/ucf-crime/experiments/frame_level/r3d_binary_20251201_173619_20251201_200629"

def main():
    results_dir = Path(PATH_TO_RESULTS)
    plot_dir = results_dir / "plots"

    if not plot_dir.exists():
        print(f"Plot directory not found: {plot_dir}")
        sys.exit(1)

    # Find all PNG files in the plot directory
    plot_files = sorted(plot_dir.glob("*.png"))

    if not plot_files:
        print(f"No plots found in: {plot_dir}")
        sys.exit(1)

    print(f"Found {len(plot_files)} plots")

    # Display each plot
    for plot_file in plot_files:
        print(f"\nShowing: {plot_file.name}")
        img = mpimg.imread(plot_file)
        
        plt.figure(figsize=(12, 8))
        plt.imshow(img)
        plt.axis('off')
        plt.title(plot_file.stem, fontsize=14)
        plt.tight_layout()
        plt.show()

if __name__ == "__main__":
    main()
```
